## Data Visualization

```python
df["Month"] = df["Date"].dt.strftime("%B")

monthly_sales = df.groupby("Month")["Total Amount"].sum()

monthly_sales = monthly_sales.reindex([
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
])

plt.figure(figsize=(12,5))

plt.plot(monthly_sales.index, monthly_sales.values, marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.show()
```

## Insight 1 - Monthly Sales Trend

The monthly sales trend shows that sales vary throughout the year.

- The highest sales were recorded in **May**.
- The lowest sales were recorded in **September**.
- Sales increased again during **October** and **December**.

This indicates seasonal fluctuations in customer purchases. The business can prepare additional inventory and marketing campaigns during high-sales months to maximize revenue.

```python
top10 = df.nlargest(10, "Total Amount")

plt.figure(figsize=(10,5))

plt.bar(
    top10["Transaction ID"].astype(str),
    top10["Total Amount"]
)

plt.title("Top 10 Highest Sales Transactions")
plt.xlabel("Transaction ID")
plt.ylabel("Total Amount")

plt.xticks(rotation=45)

plt.show()
```

## Insight 2 - Top 10 Highest Sales Transactions

The chart shows the **Top 10 highest sales transactions** based on the total transaction amount.

- The chart displays the top 10 transactions with the highest sales amount.
- All these transactions have total sales close to ₹2000, indicating high-value purchases.
- These transactions likely involved either expensive products or larger quantities.

This indicates that these transactions contributed significantly to the overall revenue. Analyzing these high-value purchases can help the business understand customer buying behavior and develop strategies to increase high-value sales.

```python
quantity = df.groupby("Product Category")["Quantity"].sum()

plt.figure(figsize=(8,5))

plt.bar(quantity.index, quantity.values)

plt.title("Quantity Sold by Product Category")
plt.xlabel("Category")
plt.ylabel("Quantity")

plt.show()
```

## Insight 3 - Quantity Sold by Product Category

The chart shows the total quantity of products sold across different product categories.

- **Clothing** recorded the highest quantity sold.
- **Electronics** was the second highest-selling category.
- **Beauty** had the lowest quantity sold among the three categories.

This indicates that customers purchased more clothing products compared to other categories. The business should ensure sufficient inventory for clothing products while exploring strategies to increase sales in the Beauty category.

```python
import seaborn as sns

plt.figure(figsize=(8,5))

sns.heatmap(
    df.select_dtypes(include=["number"]).corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

## Insight 4 - Correlation Heatmap

The correlation heatmap shows the relationship between numerical variables in the retail sales dataset.

- **Price per Unit** and **Total Amount** have a strong positive correlation (**0.85**).
- **Quantity** has a moderate positive correlation (**0.37**) with **Total Amount**.
- **Age** and **Transaction ID** have very weak or no significant correlation with other variables.

This indicates that the total sales amount is mainly influenced by the product price, while customer age and transaction order have little impact on sales performance.

```python
plt.figure(figsize=(8,5))

sns.boxplot(y=df["Total Amount"])

plt.title("Box Plot of Total Sales")

plt.show()
```

## Insight 5 - Box Plot of Total Sales

The box plot shows the distribution of total sales amounts across all transactions.

- Most sales transactions fall within a moderate sales range.
- The median sales amount is relatively low compared to the maximum value.
- A few transactions have significantly higher sales amounts, indicating high-value purchases.

This suggests that while most customers make moderate purchases, a small number of high-value transactions contribute substantially to the overall revenue. The business can focus on encouraging more high-value purchases through targeted offers and promotions.

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Age"], df["Total Amount"])

plt.title("Age vs Total Amount")
plt.xlabel("Age")
plt.ylabel("Total Amount")

plt.show()
```

## Insight 6 - Age vs Total Amount

The scatter plot shows the relationship between customer age and the total purchase amount.

- Customers of different age groups made purchases across a wide range of sales amounts.
- There is no clear upward or downward trend between age and total sales.
- High-value transactions are distributed across multiple age groups rather than being concentrated in a specific age range.

This indicates that customer age has little influence on the total purchase amount. Therefore, marketing strategies can target customers of all age groups instead of focusing on a single age category.

```python
plt.figure(figsize=(6,5))

sns.countplot(x="Gender", data=df)

plt.title("Customer Gender Distribution")

plt.show()
```

## Insight 7 - Customer Gender Distribution

The customer gender distribution is nearly balanced between male and female customers.

- Female customers are slightly more than male customers.
- Both genders contribute significantly to the total customer base.
- The small difference indicates that the business attracts customers from both genders almost equally.

This suggests that marketing campaigns and product offerings can be designed to target both male and female customers, as neither group dominates the customer base.

```python
avg_sales = df.groupby("Product Category")["Total Amount"].mean()

plt.figure(figsize=(8,5))

plt.bar(avg_sales.index, avg_sales.values)

plt.title("Average Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Average Sales")

plt.show()
```

## Insight 8 - Average Sales by Product Category

The average sales across the three product categories are nearly similar.

- **Beauty** products have the highest average sales.
- **Electronics** follows closely with a similar average.
- **Clothing** has the lowest average sales among the three categories.

This indicates that customer spending is relatively balanced across product categories, with Beauty products generating slightly higher average sales.
